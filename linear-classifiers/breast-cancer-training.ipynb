After installing the libraries, we'll start executing the code block below. 

* `bc.train.0` contains the training data. `numpy` is used to load the data 
* `bc.test.0` contains the test data.

````python
import numpy as np

f = open("bc.train.0")
# loads the 2d matrix 
data = np.loadtxt(f)

# Select all rows, and select all columns starting from index 1 to the end, skipping labels
train = data[:,1:] 

# labels are on the first column of the dataset 
train_labels = data[:,0]


# open and load test data 
f = open("bc.test.0")
data = np.loadtxt(f)

# Select all rows, and select all columns starting from index 1 to the end, skipping labels
test = data[:,1:]

# labels are on the first column of the dataset 
test_labels = data[:,0]
````

### Run the Difference in Error Predictions

#### How to read the data
It shows you how well a linear model can separate the two classes using only the given features, on data it has never seen.

* **Low error rate**: the classes are roughly linearly separable and the features carry real predictive signal.
* **High error rate**: the relationship is probably non-linear, the features aren't informative enough, or the model needs tuning (e.g., feature scaling, a different C, or a non-linear kernel).

````python
# Python function to calculate errors as a %
def error_predictions(prediction, test_labels, val_c):
    err = 0
    for i in range(0, len(prediction), 1):
        if(prediction[i] != test_labels[i]):
            err += 1
    err = err/len(test_labels)
    print(f"errors: " + str(err) + " | percentage: " + str(err*100) + "% | C: " + str(val_c))
````

### Least Squares 

Use `LinearRegression` with least squares

````python
from sklearn.linear_model import LinearRegression

clf = LinearRegression()
clf.fit(train, train_labels)
prediction = clf.predict(test)

# reduces data to -1 and 1 
prediction = np.sign(prediction)

# print("predictions: ", prediction)
# print("test_labels: ", test_labels)

error_predictions(prediction, test_labels, "N/A")
````

### Support Vector Machines

Running `LinearSVC` (Linear Support Vector Classification), and printing the predictions

* [LinearSVC API Doc](https://scikit-learn.org/stable/modules/generated/sklearn.svm.LinearSVC.html)
* [Scaling the regularization parameter for SVCs](https://scikit-learn.org/stable/auto_examples/svm/plot_svm_scale_c.html#sphx-glr-auto-examples-svm-plot-svm-scale-c-py)

````python
from sklearn.svm import LinearSVC

# default is 0.0001

# reducing the tol increases error percentage.
# 0.001 error is 1.75% & 0.0001 error is 3.44%

# Play around with C, where cross validation comes in

Cs =  [0.01, 0.1, 1, 10, 100]

print(" errors   |  percentage  |   C  ")
print(" -------- | ------------ | ------ ")
for C in Cs: 
    clf = LinearSVC(dual=False, tol=0.0001, C=C)
    clf.fit(train, train_labels)
    prediction = clf.predict(test)

    # print("predictions: ", prediction)
    # print("test_labels: ", test_labels)

    error_predictions(prediction, test_labels, C)
````

To get the true error you need to do class wise. Check for error in each class. This can be validated using `cross_val_score`.

````python
from sklearn.model_selection import cross_val_score

# combine train and test
combined_data = np.concatenate((train, test))
combined_labels = np.concatenate((train_labels, test_labels))

five_fold_accuracy = cross_val_score(clf, combined_data, combined_labels, scoring="balanced_accuracy")
print("Five Fold Accuracy: ", five_fold_accuracy)
print("Accuracy Mean: ", np.mean(five_fold_accuracy))
````
